```python
#Sistema de atendimento de uma clínica
#usando lista encadeada (linked list) em Python.


# 18/19 - Paciente e Nó da lista encadeada

class Paciente:
    def __init__(self, nome, idade, prioridade="Normal"):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade  # "Normal" ou "Prioridade"

    def __str__(self):
        return f"{self.nome} - {self.idade} anos - {self.prioridade}"


class Node:
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


# 20/21 - Fila de atendimento (lista encadeada simples)

class FilaAtendimento:
    def __init__(self):
        self.inicio = None
        self.fim = None

    def adicionar(self, paciente):
        """Adiciona um paciente no fim da fila."""
        novo_no = Node(paciente)

        if self.esta_vazia():
            # fila vazia: novo nó é início e fim
            self.inicio = novo_no
            self.fim = novo_no
        else:
            # liga o antigo fim ao novo nó
            self.fim.proximo = novo_no
            self.fim = novo_no

    def atender(self):
        """Remove e retorna o primeiro paciente da fila."""
        if self.esta_vazia():
            print("Fila vazia, ninguém para atender.")
            return None

        no_atendido = self.inicio
        self.inicio = self.inicio.proximo

        # se a fila ficou vazia, fim também precisa ser resetado
        if self.inicio is None:
            self.fim = None

        return no_atendido.paciente

    def listar(self):
        """Mostra todos os pacientes da fila, em ordem."""
        if self.esta_vazia():
            print("Fila vazia.")
            return

        atual = self.inicio
        posicao = 1
        while atual is not None:
            print(f"{posicao}. {atual.paciente}")
            atual = atual.proximo
            posicao += 1

    def esta_vazia(self):
        """Retorna True se a fila não tiver pacientes."""
        return self.inicio is None

    def tamanho(self):
        """Retorna a quantidade de pacientes na fila."""
        contador = 0
        atual = self.inicio
        while atual is not None:
            contador += 1
            atual = atual.proximo
        return contador


# 22 - Exemplo de utilização

if __name__ == "__main__":
    print("=== Exemplo simples (sem prioridade) ===")
    fila = FilaAtendimento()

    fila.adicionar(Paciente("Ana", 32, "Normal"))
    fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
    fila.adicionar(Paciente("Carlos", 45, "Normal"))

    fila.listar()

    paciente = fila.atender()
    print("Atendendo:", paciente.nome)

    print("Tamanho da fila:", fila.tamanho())
    print("Fila vazia?", fila.esta_vazia())

    # 23 - Desafio: atendimento prioritário


    print("\n=== Desafio: atendimento prioritário ===")

    class FilaPrioritaria(FilaAtendimento):
        def adicionar(self, paciente):
            novo_no = Node(paciente)

            # caso 1: fila vazia
            if self.esta_vazia():
                self.inicio = novo_no
                self.fim = novo_no
                return

            # caso 2: paciente é prioridade -> insere após o último
            # paciente de prioridade (ou no início, se não houver nenhum)
            if paciente.prioridade == "Prioridade":
                anterior = None
                atual = self.inicio
                while atual is not None and atual.paciente.prioridade == "Prioridade":
                    anterior = atual
                    atual = atual.proximo

                if anterior is None:
                    # vira o novo início
                    novo_no.proximo = self.inicio
                    self.inicio = novo_no
                else:
                    novo_no.proximo = anterior.proximo
                    anterior.proximo = novo_no
                    if novo_no.proximo is None:
                        self.fim = novo_no
            else:
                # caso 3: paciente normal -> vai para o fim, como sempre
                self.fim.proximo = novo_no
                self.fim = novo_no

    fila2 = FilaPrioritaria()
    fila2.adicionar(Paciente("Ana", 32, "Normal"))
    fila2.adicionar(Paciente("Bruno", 70, "Normal"))
    fila2.adicionar(Paciente("Carlos", 45, "Prioridade"))

    print("Ordem esperada: 1. Carlos  2. Ana  3. Bruno\n")
    print("Ordem obtida:")
    fila2.listar()
```
