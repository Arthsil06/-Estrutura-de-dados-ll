```python
# EXERCÍCIO 1 - Pilha (Simulador Desfazer)

pilha = []

def digitar(texto):
    pilha.append(texto)
    print("Digitou:", texto)

def desfazer():
    if pilha:
        removido = pilha.pop()
        print("Desfez:", removido)
    else:
        print("Nada para desfazer")

print("--- Exercício 1: Pilha ---")
digitar("Ola")
digitar(" mundo")
digitar("!")
desfazer()
desfazer()
desfazer()
desfazer()


# EXERCÍCIO 2 - Fila (Sistema de Impressão)

fila = []

def enfileirar(documento):
    fila.append(documento)
    print("Adicionou na fila:", documento)

def imprimir():
    if fila:
        doc = fila.pop(0)
        print("Imprimindo:", doc)
    else:
        print("Fila vazia")

print("\n--- Exercício 2: Fila ---")
enfileirar("Relatorio.pdf")
enfileirar("Boleto.pdf")
enfileirar("Contrato.docx")
imprimir()
imprimir()
imprimir()
imprimir()
```
