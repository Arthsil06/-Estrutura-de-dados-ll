```python
from dataclasses import dataclass


# Atividade 1  Listas
def atividade1():
    print("\n--- Atividade 1: Listas ---")
    numeros = []
    for i in range(10):
        numeros.append(int(input(f"Digite o {i+1}º numero: ")))

    print("Numeros:", numeros)
    soma = sum(numeros)
    media = soma / len(numeros)
    print("Soma:", soma)
    print("Media:", media)

    # Desafio
    print("Maior:", max(numeros))
    print("Menor:", min(numeros))
    pares = [n for n in numeros if n % 2 == 0]
    print("Quantidade de pares:", len(pares))


#Atividade 2  Matrizes
def atividade2():
    print("\n--- Atividade 2: Matrizes ---")
    matriz = []
    for i in range(3):
        linha = []
        for j in range(3):
            linha.append(int(input(f"Elemento [{i}][{j}]: ")))
        matriz.append(linha)

    print("Matriz:")
    for linha in matriz:
        print(linha)

    soma_total = sum(sum(linha) for linha in matriz)
    diagonal = sum(matriz[i][i] for i in range(3))
    maior = max(max(linha) for linha in matriz)

    print("Soma total:", soma_total)
    print("Soma diagonal principal:", diagonal)
    print("Maior elemento:", maior)


#Atividade 3  Dicionario
def atividade3():
    print("\n--- Atividade 3: Dicionário ---")
    livro = {
        "titulo": input("Título: "),
        "autor": input("Autor: "),
        "ano": input("Ano: "),
        "preco": float(input("Preço: ")),
    }

    print("Dados do livro:", livro)

    livro["preco"] = float(input("Novo preço: "))
    livro["categoria"] = input("Categoria: ")

    print("Livro atualizado:", livro)


#  Atividade 4 e 5  Dataclass Aluno
@dataclass
class Aluno:
    nome: str
    matricula: int
    nota1: float
    nota2: float

    def media(self):
        return (self.nota1 + self.nota2) / 2


def atividade5():
    print("\n--- Atividade 5: Lista de Alunos ---")
    alunos = []
    for i in range(5):
        print(f"\nAluno {i+1}:")
        nome = input("Nome: ")
        matricula = int(input("Matrícula: "))
        nota1 = float(input("Nota 1: "))
        nota2 = float(input("Nota 2: "))
        alunos.append(Aluno(nome, matricula, nota1, nota2))

    print("\nMédias:")
    for a in alunos:
        print(f"{a.nome}: {a.media():.2f}")

    print("\nAprovados (média >= 7):")
    for a in alunos:
        if a.media() >= 7:
            print(a.nome)

    melhor = max(alunos, key=lambda a: a.media())
    print(f"\nMaior média: {melhor.nome} ({melhor.media():.2f})")


#Estoque
@dataclass
class Produto:
    codigo: int
    nome: str
    preco: float
    quantidade: int


def sistema_estoque():
    estoque = []

    while True:
        print("""
===== SISTEMA DE ESTOQUE =====
1 - Cadastrar produto
2 - Listar produtos
3 - Buscar produto
4 - Valor total do estoque
5 - Produto mais caro
0 - Sair
""")
        opcao = input("Escolha: ")

        if opcao == "1":
            codigo = int(input("Código: "))
            nome = input("Nome: ")
            preco = float(input("Preço: "))
            quantidade = int(input("Quantidade: "))
            estoque.append(Produto(codigo, nome, preco, quantidade))
            print("Produto cadastrado!")

        elif opcao == "2":
            if not estoque:
                print("Estoque vazio.")
            for p in estoque:
                print(p)

        elif opcao == "3":
            codigo = int(input("Código a buscar: "))
            encontrado = next((p for p in estoque if p.codigo == codigo), None)
            print(encontrado if encontrado else "Produto não encontrado.")

        elif opcao == "4":
            total = sum(p.preco * p.quantidade for p in estoque)
            print(f"Valor total do estoque: R$ {total:.2f}")

        elif opcao == "5":
            if estoque:
                mais_caro = max(estoque, key=lambda p: p.preco)
                print("Produto mais caro:", mais_caro)
            else:
                print("Estoque vazio.")

        elif opcao == "0":
            print("Saindo...")
            break

        else:
            print("Opção inválida.")


#Menu
if __name__ == "__main__":
    print("Escolha a atividade para executar:")
    print("1 - Listas")
    print("2 - Matrizes")
    print("3 - Dicionário")
    print("5 - Lista de Alunos (inclui Atividade 4)")
    print("6 - Sistema de Estoque")

    escolha = input("Opção: ")

    if escolha == "1":
        atividade1()
    elif escolha == "2":
        atividade2()
    elif escolha == "3":
        atividade3()
    elif escolha == "5":
        atividade5()
    elif escolha == "6":
        sistema_estoque()
    else:
        print("Opção inválida.")
```
